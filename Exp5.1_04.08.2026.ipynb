```python
import pandas as pd
import numpy as np

# Create a large dataset (10,000 rows)
np.random.seed(42)

df = pd.DataFrame({
    "Student_ID": range(1, 10001),
    "Age": np.random.randint(18, 30, 10000),
    "Marks": np.random.randint(0, 101, 10000),
    "Salary": np.random.randint(20000, 150001, 10000)
})

print(df.head())
```

```python
# Divide Marks into 5 equal-width bins
df["Marks_Bin"] = pd.cut(df["Marks"], bins=5)

print(df[["Marks", "Marks_Bin"]].head(10))
```

```python
labels = ["Very Low", "Low", "Medium", "High", "Very High"]

df["Marks_Category"] = pd.cut(
    df["Marks"],
    bins=5,
    labels=labels
)

print(df[["Marks", "Marks_Category"]].head(10))
```

```python
df["Salary_Bin"] = pd.qcut(
    df["Salary"],
    q=4,
    labels=["Q1", "Q2", "Q3", "Q4"]
)

print(df[["Salary", "Salary_Bin"]].head(10))
```

```python
print("\nMarks Bin Count:")
print(df["Marks_Bin"].value_counts())

print("\nSalary Bin Count:")
print(df["Salary_Bin"].value_counts())
```
