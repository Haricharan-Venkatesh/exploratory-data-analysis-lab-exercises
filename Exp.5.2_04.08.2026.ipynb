```python
import pandas as pd
import numpy as np

np.random.seed(42)

# Create dataset with 10,000 rows
df = pd.DataFrame({
    "Age": np.random.randint(18, 60, 10000),
    "Salary": np.random.randint(20000, 150001, 10000),
    "Experience": np.random.randint(0, 41, 10000)
})

print(df.head())
```

```python
from sklearn.preprocessing import MinMaxScaler

minmax = MinMaxScaler()

df_minmax = pd.DataFrame(
    minmax.fit_transform(df),
    columns=df.columns
)

print("Min-Max Scaled Data:")
print(df_minmax.head())
```

```python
from sklearn.preprocessing import StandardScaler

standard = StandardScaler()

df_standard = pd.DataFrame(
    standard.fit_transform(df),
    columns=df.columns
)

print("Standardized Data:")
print(df_standard.head())
```

```python
from sklearn.preprocessing import RobustScaler

robust = RobustScaler()

df_robust = pd.DataFrame(
    robust.fit_transform(df),
    columns=df.columns
)

print("Robust Scaled Data:")
print(df_robust.head())
```
