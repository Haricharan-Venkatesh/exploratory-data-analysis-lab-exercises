```python
print("25BDS0228\nHaricharan Venkatesh")

n = int(input("Enter a number: "))

if n == 1:
    print("1 is not a prime number")
else:
    c = 0
    for i in range(2, n):
        if n % i == 0:
            c += 1

    if c == 0:
        print(f"{n} is a prime number")
    else:
        print(f"{n} is not a prime number")
```

```python
print("25BDS0228\nHaricharan Venkatesh")

n = int(input("Enter a number: "))
r = 0
num = n

while num:
    r = r * 10 + num % 10
    num //= 10

if n == r:
    print(f"{n} is a palindrome")
else:
    print(f"{n} is not a palindrome")
```

```python
print("25BDS0228\nHaricharan Venkatesh")

array = [1, 23, 45, 67, 78]
target = int(input(f"Enter a number to search in a sorted array {array}: "))

l = 0
r = len(array) - 1

while l <= r:
    mid = (l + r) // 2

    if array[mid] == target:
        print(f"{target} found in array {array} in index {mid}")
        break
    elif target > array[mid]:
        l = mid + 1
    else:
        r = mid - 1
else:
    print(f"{target} not found in array {array}")
```
