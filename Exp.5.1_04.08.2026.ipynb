```python
import pandas as pd
import numpy as np

# Create a sample dataset
np.random.seed(42)

df = pd.DataFrame({
    "Student": [f"S{i}" for i in range(1, 21)],
    "Marks": np.random.randint(35, 101, 20)
})

print("Original Dataset:")
print(df)
```

```python
# Define bins
bins = [30, 50, 70, 90, 100]
labels = ["Poor", "Average", "Good", "Excellent"]

df["Equal_Width_Bin"] = pd.cut(df["Marks"], bins=bins, labels=labels)

print("\nEqual Width Binning:")
print(df)

print("\nCount in each bin:")
print(df["Equal_Width_Bin"].value_counts())
```

```python
df["Equal_Frequency_Bin"] = pd.qcut(
    df["Marks"],
    q=4,
    labels=["Q1", "Q2", "Q3", "Q4"]
)

print("\nEqual Frequency Binning:")
print(df)

print("\nCount in each frequency bin:")
print(df["Equal_Frequency_Bin"].value_counts())
```
